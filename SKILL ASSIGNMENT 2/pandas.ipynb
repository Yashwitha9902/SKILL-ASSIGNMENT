Question 1 : Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

```python
import pandas as pd
import numpy as np

data = np.random.randint(1, 10, (5, 3))

df = pd.DataFrame(data, columns=["A", "B", "C"])

df["Product"] = df["A"] * df["B"]

print(df)
```

Question 2 : Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

```python


data = np.random.randint(1, 10, (4, 3))

df = pd.DataFrame(data, columns=["A", "B", "C"])

print(df)

print("Row-wise sum:")
print(df.sum(axis=1))

print("Column-wise sum:")
print(df.sum(axis=0))
```

Question 3 :  Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python


data = np.random.randint(1, 10, (5, 3))

df = pd.DataFrame(data, columns=["A", "B", "C"])

df.iloc[1, 0] = np.nan
df.iloc[3, 1] = np.nan
df.iloc[2, 2] = np.nan

print("Original DataFrame:")
print(df)

df = df.fillna(df.mean())

print("DataFrame after filling NaN:")
print(df)
```

Question 4 : Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

```python
data = np.random.randint(1, 10, (6, 4))

df = pd.DataFrame(data, columns=["A", "B", "C", "D"])

df.iloc[1, 0] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 1] = np.nan

print("Original DataFrame:")
print(df)

df = df.dropna()

print("DataFrame after dropping rows:")
print(df)
```

Question 5 : Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.


```python
data = {
    "Category": np.random.choice(["A", "B", "C"], 10),
    "Value": np.random.randint(1, 100, 10)
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

result = df.groupby("Category")["Value"].agg(["sum", "mean"])

print("Grouped Data:")
print(result)
```

Question 6 :  Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
data = {
    "Product": ["Pen", "Book", "Pencil", "Bag", "Scale"],
    "Category": ["A", "B", "A", "C", "B"],
    "Sales": np.random.randint(100, 500, 5)
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

result = df.groupby("Category")["Sales"].sum()

print("Total Sales by Category:")
print(result)
```

Question 7 : Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
data1 = {
    "ID": [1, 2, 3],
    "Name": ["Asha", "Ravi", "John"]
}

data2 = {
    "ID": [1, 2, 3],
    "Marks": [85, 90, 75]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

print("First DataFrame:")
print(df1)

print("Second DataFrame:")
print(df2)

result = pd.merge(df1, df2, on="ID")

print("Merged DataFrame:")
print(result)
```

Question 8 : Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
data = {
    "Marks": [85, 90, 75, 80],
    "Age": [20, 21, 19, 20]
}

index = [
    ["A", "A", "B", "B"],
    ["Student1", "Student2", "Student3", "Student4"]
]

df = pd.DataFrame(data, index=index)

df.index.names = ["Group", "Student"]

print("Original DataFrame:")
print(df)

print("Accessing Group A:")
print(df.loc["A"])

print("Accessing Student3:")
print(df.loc[("B", "Student3")])

print("Slicing Group A:")
print(df.loc["A":"A"])
```

Question 9 : Create a Pandas DataFrame with MultiIndex consisting of Category and SubCategory. Fill the DataFrame with random data and compute the sum of values for each Category and SubCategory.

```python
index = pd.MultiIndex.from_tuples([
    ("A", "A1"),
    ("A", "A2"),
    ("B", "B1"),
    ("B", "B2")
], names=["Category", "SubCategory"])

df = pd.DataFrame({
    "Value": np.random.randint(1, 100, 4)
}, index=index)

print("Original DataFrame:")
print(df)

print("Sum of values:")
print(df.groupby(["Category", "SubCategory"])["Value"].sum())
```

Question 10 : Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
data = np.random.randint(1, 10, (5, 3))

df = pd.DataFrame(data, columns=["A", "B", "C"])

print("Original DataFrame:")
print(df)

result = df.apply(lambda x: x * 2, axis=0)

print("DataFrame after doubling:")
print(result)
```

Question 11 : Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python

data = np.random.randint(1, 10, (6, 3))

df = pd.DataFrame(data, columns=["A", "B", "C"])

print("Original DataFrame:")
print(df)

df["Total"] = df.apply(lambda x: x["A"] + x["B"] + x["C"], axis=1)

print("DataFrame with Total:")
print(df)
```

Question 12 : Create a Pandas Series with 5 random text strings. Convert all the strings to uppercase.

```python
data = ["python", "java", "machine learning", "data science", "pandas"]

s = pd.Series(data)

print("Original Series:")
print(s)

result = s.str.upper()

print("Uppercase Series:")
print(result)
```
