## NUMPY ASSIGNMENT:
Question 1: Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.


```python
import numpy as np

a = np.random.randint(1, 21, (5, 5))
print("Original array:")
print(a)
a[:, 2] = 1
print("Updated array:")
print(a)
```

Question 2: Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.


```python
a = np.arange(1, 17).reshape(4, 4)
print("Original array:")
print(a)
np.fill_diagonal(a, 0)
print("Updated array:")
print(a)

```

Question 3:  Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.


```python
a = np.random.randint(1, 21, (5, 5))
print("Original array:");
print(a)

print("Border elements:")

print("Top row:", a[0, :])

print("Bottom row:", a[-1, :])

print("Left side:", a[1:-1, 0])

print("Right side:", a[1:-1, -1])
```

Question 4 : Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

```python
a = np.arange(1, 17).reshape(4, 4)
print("Original array:")
print(a)
print("Row-wise sum:")
print(np.sum(a, axis=1))
print("Column-wise sum:")
print(np.sum(a, axis=0))
```

Question 5 : Create a NumPy array of shape (3, 3) with values from 1 to 9. Reshape the array to shape (1, 9) and then to shape (9, 1).


```python
a = np.arange(1, 10).reshape(3, 3)
print("Original array:")
print(a)
b = a.reshape(1, 9)
print("Shape (1, 9):")
print(b)
c = a.reshape(9, 1)
print("Shape (9, 1):")
print(c)

```

Question 6 :  Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float). Add some data and sort the array by age.

```python
data = np.array([
    ("Asha", 20, 50.5),

    ("Ravi", 22, 65.0),

    ("John", 19, 55.5)
], dtype=[

    ("name", "U10"),

    ("age", "i4"),

    ("weight", "f4")

])
print("Original array:")
print(data)
result = np.sort(data, order="age")
print("Array sorted by age:")
print(result)

```
