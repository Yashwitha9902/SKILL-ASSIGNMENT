## NUMPY ASSIGNMENT:
Question 1: Create a 5 × 5 NumPy array with random integers between 1 and 20. Replace all elements in the third column with 1.

```python
import numpy as np

a = np.random.randint(1, 21, (5, 5))

print("Original array:")
print(a)

a[:, 2] = 1

print("Updated array:")
print(a)
```

Question 2: Create a 4 × 4 NumPy array with values from 1 to 16 and replace the diagonal elements with 0.

```python
import numpy as np

a = np.arange(1, 17).reshape(4, 4)

print("Original array:")
print(a)

np.fill_diagonal(a, 0)

print("Updated array:")
print(a)
```

Question 3:  Find the maximum and minimum elements in a NumPy array.

```python
import numpy as np

a = np.array([10, 25, 5, 40, 15])

print("Array:", a)
print("Maximum element:", np.max(a))
print("Minimum element:", np.min(a))
```

Question 4 : Find the sum of all elements in a NumPy array

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Array:", a)
print("Sum:", np.sum(a))
```

Question 5 : Find the mean, median, and standard deviation of a NumPy array.

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Standard deviation:", np.std(a))
```

Question 6 : NumPy,  Reshape an Array

```python
import numpy as np

a = np.arange(1, 13)

print("Original array:")
print(a)

b = a.reshape(3, 4)

print("Reshaped array:")
print(b)
```
