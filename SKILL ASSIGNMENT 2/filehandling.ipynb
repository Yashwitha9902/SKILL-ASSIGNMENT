Question 1: Write a function that reads the contents of a file named document.txt and returns the number of words in the file.

```python
def count_words():
    f = open("document.txt", "r")
    data = f.read()
    words = data.split()
    print("Number of words:", len(words))
    f.close()

count_words()
```

Question 2: Write a function that finds and replaces all occurrences of a given word in a file named data.txt with another word.


```python

def replace_word():
    f = open("data.txt", "r")
    data = f.read()
    f.close()

    data = data.replace("Python", "Java")

    f = open("data.txt", "w")
    f.write(data)
    f.close()

replace_word()
print("Word replaced successfully")
```

Question 3 : Write a function that reads the contents of a file named reverse.txt and prints each line in reverse order.


```python

def reverse_lines():
    f = open("reverse.txt", "r")
    lines = f.readlines()
    f.close()

    lines.reverse()

    for line in lines:
        print(line.strip())

reverse_lines()
```

Question 4: Write a function that reads the contents of a file named stats.txt and returns the number of lines, words, and characters in the file.

```python

def file_statistics():
    f = open("stats.txt", "r")
    data = f.read()
    f.close()

    lines = data.splitlines()
    words = data.split()
    characters = len(data)

    print("Number of lines:", len(lines))
    print("Number of words:", len(words))
    print("Number of characters:", characters)

file_statistics()
```
